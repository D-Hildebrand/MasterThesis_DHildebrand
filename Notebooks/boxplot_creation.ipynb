```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
filelist = ["alpha_div_family.csv", "alpha_div_genus.csv", "alpha_div_kingdom.csv"]
```

```python
df = pd.DataFrame()
for file in filelist: 
    phylum_df = pd.read_csv(file, usecols= [r"%unclassified"])
    phylum_df.rename(columns={r"%unclassified" : file.split("_")[-1].split(".")[0]}, 
                     inplace=True)
    df = phylum_df.join(df)
df
```

```python
sns.boxplot(df)
plt.xticks(rotation=45)
plt.ylabel("% Assigned")
plt.xlabel("Assignment level")
plt.title("Boxplot of percentage of Kaiju classification")
plt.ylim(0,100)



# plt.savefig("testimg.png")
# plt.savefig("testimg.svg")
```
